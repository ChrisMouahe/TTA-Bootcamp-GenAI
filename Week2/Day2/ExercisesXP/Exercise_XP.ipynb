Exercice 6 : Encodage des fonctionnalités
But : Finaliser l’encodage des variables catégorielles.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.impute import SimpleImputer

#Gestion des valeurs manquantes
titanic_data = pd.read_csv('/content/titanic dataset/train.csv')
imputer = SimpleImputer(strategy='most_frequent')
titanic_data = pd.DataFrame(imputer.fit_transform(titanic_data), columns=titanic_data.columns)
numeric_columns = ['Age', 'Fare', 'SibSp', 'Parch']

titanic_data[numeric_columns] = titanic_data[numeric_columns].astype(float)
titanic_data

#gestion des valeurs aberrantes

#Détection des valeurs aberrantes de Age par la méthode IQR

Q1 = titanic_data['Age'].quantile(0.25)
Q3 = titanic_data['Age'].quantile(0.75)

IQR = Q3 - Q1

lower_bound_age = Q1 - 1.5 * IQR
upper_bound_age = Q3 + 1.5 * IQR

outliers_age = titanic_data[(titanic_data['Age'] < lower_bound_age) | (titanic_data['Age'] > upper_bound_age)]
outliers_age[['Age']]

#gestion des valeurs aberrantes pour Age avec plafond de quantile à 0,98
cap_value = titanic_data['Age'].quantile(0.98)

titanic_data['Age_capped'] = np.where(
    titanic_data['Age'] > cap_value,
    cap_value,
    titanic_data['Age']
)

#Transformation logarithmique de Age
titanic_data['Age_log'] = np.log1p(titanic_data['Age'])

#Suppression des lignes aberrantes pour Age
no_outliers = titanic_data[
    (titanic_data['Age'] >= lower_bound_age) &
    (titanic_data['Age'] <= upper_bound_age)
]

#Détection des valeurs aberrantes de are par la méthode IQR

Q1 = titanic_data['Fare'].quantile(0.25)
Q3 = titanic_data['Fare'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers_fare = titanic_data[(titanic_data['Fare'] < lower_bound) | (titanic_data['Fare'] > upper_bound)]
outliers_fare[['Fare']]

#gestion des valeurs aberrantes pour Fare avec plafond de quantile à 0,98
cap_value = titanic_data['Fare'].quantile(0.98)

titanic_data['Fare_capped'] = np.where(
    titanic_data['Fare'] > cap_value,
    cap_value,
    titanic_data['Fare']
)

#Transformation logarithmique de Fare
titanic_data['Fare_log'] = np.log1p(titanic_data['Fare'])

#Suppression des lignes aberrantes pour Fare
no_outliers = titanic_data[
    (titanic_data['Fare'] >= lower_bound) &
    (titanic_data['Fare'] <= upper_bound)
]
#Retrait du titre
no_outliers = no_outliers.copy()
no_outliers['Title'] = titanic_data['Name'].str.extract(
    ' ([A-Za-z]+)\.',
    expand=False
)
no_outliers = pd.get_dummies(
    no_outliers,
    columns=['Sex', 'Embarked', 'Title'],
    drop_first=True
)

#Encodage de Age
standard_scaler = StandardScaler()

no_outliers['Age_scaled'] = standard_scaler.fit_transform(
    no_outliers[['Age']]
)

#Encodadge de Fare
minmax_scaler = MinMaxScaler()

no_outliers['Fare_scaled'] = minmax_scaler.fit_transform(
    no_outliers[['Fare']]
)
no_outliers.head()
```

Exercice 7 : Transformation des données pour l’âge
But : Créez et encodez des groupes d’âge.

```python
import pandas as pd

titanic_data = pd.read_csv('/content/titanic dataset/train.csv')
bins = [0, 12, 18, 60, 100]

labels = ['Child', 'Teen', 'Adult', 'Senior']

titanic_data['Age_Group'] = pd.cut(
    titanic_data['Age'],
    bins=bins,
    labels=labels
)
titanic_data = pd.get_dummies(
    titanic_data,
    columns=['Age_Group'],
    drop_first=True
)
titanic_data.head()
#titanic_data.columns
```
