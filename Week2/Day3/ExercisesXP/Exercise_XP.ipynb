Exercice 2 : Statistiques descriptives
Tâche : Étant donné un ensemble de données d'exemple, calculez la moyenne, la médiane, la variance et l'écart type à l'aide de SciPy.
Exemple de jeu de données :
data = [12, 15, 13, 12, 18, 20, 22, 21]

```python
data = [12, 15, 13, 12, 18, 20, 22, 21]
#calcul de la moyenne
mean = stats.tmean(data)
print(mean) 
#Calcul de la médiane
median = np.median(data)
print(median)
#Calcul de l'écart type
std_dev = stats.tstd(data)
print(std_dev)
#calcul de la variance
variance = stats.tvar(data)
print(variance)
```

Exercice 3 : Comprendre les distributions
Tâche : Générer une distribution normale à l'aide de SciPy avec une moyenne de 50 et un écart type de 10. Représenter graphiquement cette distribution.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mu, sigma = 50, 10
data = stats.norm(mu, sigma)
plt.hist(data.rvs(1000), bins=30, density=True, alpha=0.6, color='g')
plt.title('Distribution Normale')
```

Exercice 4 : Application du test t
Tâche : Effectuer un test T sur deux ensembles de données générées aléatoirement.


```python
import numpy as np
from scipy import stats


data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

t_stat, p_value = stats.ttest_ind(data1, data2)

print("Statistique t :", t_stat)
print("p_value :", p_value)
if p_value < 0.05:
    print("Les moyennes sont significativement différentes.")
else:
    print("Aucune différence significative entre les moyennes.")
```

Exercice 5 : Analyse de régression linéaire
Objectif : Appliquer la régression linéaire à un ensemble de données et interpréter les résultats.

Tâche : Étant donné un ensemble de données de prix de logements ( house_prices) et leurs tailles correspondantes ( house_sizes), utilisez la régression linéaire pour prédire le prix d'une maison en fonction de sa taille.

```python
import numpy as np
from scipy.stats import linregress

house_sizes = np.array([50, 70, 80, 100, 120])
house_prices = np.array([150000, 200000, 210000, 250000, 280000])

slope, intercept, r_value, p_value, std_err = linregress(
    house_sizes,
    house_prices
)

print("Pente :", slope)
print("Ordonnée à l'origine :", intercept)

price_90 = slope * 90 + intercept

print("Prix estimé d'une maison de 90 m² :", price_90)
```

Exercice 6 : Comprendre l’ANOVA
Objectif : Tester la compréhension de l'ANOVA et de son application.

Tâche : Trois engrais différents sont appliqués à trois groupes de plantes distincts afin de tester leur efficacité. La croissance, exprimée en centimètres, est enregistrée.


```python
from scipy import stats

fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

F_stat, p_value = stats.f_oneway(
    fertilizer_1,
    fertilizer_2,
    fertilizer_3
)

print("F =", F_stat)
print("p =", p_value)

```

Comme : p < 0.05, on rejette l'hypothèse nulle.
Les trois engrais n'ont pas le même effet sur la croissance des plantes.
Au moins un groupe possède une moyenne significativement différente des autres.

Si : p > 0.05 alors : On ne pourrait pas rejeter l'hypothèse nulle. Cela signifierait que les différences observées entre les groupes pourraient être dues au hasard.
