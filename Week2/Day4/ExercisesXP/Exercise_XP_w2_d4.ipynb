###  Exercice 1 : Comprendre la visualisation des données

**Pourquoi la visualisation des données est-elle importante ?**
La visualisation transforme des données brutes et complexes en représentations graphiques faciles à comprendre. Elle permet d'identifier rapidement des tendances, des anomalies et des corrélations qui seraient difficiles à détecter dans un simple tableau de chiffres. C'est un outil de communication essentiel pour transmettre des conclusions de manière claire et percutante.

**Utilité d'un graphique linéaire :**
Un graphique linéaire est principalement utilisé pour montrer l'évolution d'une variable au fil du temps (séries temporelles). Il permet de visualiser la continuité, la direction d'une tendance (hausse ou baisse) et la vitesse de changement entre les points de données.

```python
import matplotlib.pyplot as plt

# Données de l'exercice 2
jours = ['Lun', 'Mar', 'Mer', 'Jeu', 'Ven', 'Sam', 'Dim']
temperatures = [72, 74, 76, 80, 82, 78, 75]

# Création du graphique linéaire
plt.figure(figsize=(8, 5))
plt.plot(jours, temperatures, marker='o', linestyle='-', color='orange')

# Personnalisation
plt.title('Variation de la température sur une semaine')
plt.xlabel('Jour')
plt.ylabel('Température (°F)')
plt.grid(True, linestyle='--', alpha=0.6)

# Affichage
plt.show()
```

```python
# Données de l'exercice 3
mois = ['Jan', 'Fév', 'Mar', 'Avr', 'Mai']
ventes = [5000, 5500, 6200, 7000, 7500]

# Création du graphique à barres
plt.figure(figsize=(8, 5))
plt.bar(mois, ventes, color='skyblue')

# Personnalisation
plt.title('Ventes mensuelles du magasin')
plt.xlabel('Mois')
plt.ylabel('Montant des ventes ($)')

# Affichage
plt.show()
```

### Préparation des données pour les exercices 4, 5 et 6
Nous allons télécharger le dataset 'Student Mental Health' depuis une source publique pour pouvoir travailler sur les exercices suivants.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Comme les liens externes échouent, nous créons un échantillon de données représentatif du dataset 'Student Mental Health'
data = {
    'Choose your gender': ['Male', 'Female', 'Female', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'Age': [18, 21, 19, 22, 23, 20, 24, 18, 22, 19],
    'What is your CGPA?': ['3.00 - 3.49', '3.50 - 4.00', '3.50 - 4.00', '2.50 - 2.99', '3.00 - 3.49', '3.50 - 4.00', '3.00 - 3.49', '3.50 - 4.00', '2.50 - 2.99', '3.50 - 4.00'],
    'Do you have Anxiety?': ['No', 'Yes', 'Yes', 'No', 'No', 'No', 'Yes', 'No', 'No', 'Yes'],
    'Do you have Panic attack?': ['Yes', 'No', 'Yes', 'No', 'No', 'No', 'Yes', 'No', 'No', 'No']
}

try:
    df = pd.DataFrame(data)
    # Nettoyage des colonnes
    df.columns = [c.strip() for c in df.columns]
    print("Données générées avec succès (Données locales de secours).")
    display(df.head())
except Exception as e:
    print(f"Erreur lors de la préparation : {e}")
```

```python
# Exercice 4 : Distribution des CGPA
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='What is your CGPA?', color='teal', kde=True)
plt.title('Distribution des moyennes cumulatives (CGPA)')
plt.xticks(rotation=45)
plt.show()
```

```python
# Exercice 5 : Comparaison des niveaux d’anxiété selon le sexe
plt.figure(figsize=(8, 6))
sns.countplot(data=df, x='Choose your gender', hue='Do you have Anxiety?', palette='pastel')
plt.title('Comparaison des niveaux d’anxiété selon le sexe')
plt.xlabel('Sexe')
plt.ylabel("Nombre d'étudiants")
plt.legend(title='Souffre d\'anxiété ?')
plt.show()
```

```python
# Exercice 6 : Relation entre l’âge et les crises de panique
# Conversion des réponses en valeurs numériques
df['Panic_Numeric'] = df['Do you have Panic attack?'].map({'Yes': 1, 'No': 0})

plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Age', y='Panic_Numeric', alpha=0.5, s=100, color='red')

plt.title('Relation entre l’âge et les crises de panique')
plt.xlabel('Âge')
plt.ylabel('Crise de panique (1=Oui, 0=Non)')
plt.yticks([0, 1], ['Non', 'Oui'])
plt.grid(True, axis='x', linestyle='--', alpha=0.7)
plt.show()
```
