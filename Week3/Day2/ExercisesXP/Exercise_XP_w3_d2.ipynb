 Exercice 1 : Opérations matricielles
Instructions

Calculez le déterminant.
Trouvez l'inverse de la matrice.

```python
import pandas as pd
import numpy as np
from numpy import random
from scipy import stats
import matplotlib.pyplot as plt

array = np.array([1, 2, 4, 6, 3, 5, 9, 7, 8]).reshape(3,3)
array
```

```python
# Calcul du déterminant et de l'inverse
determinant = np.linalg.det(array)
inverse = np.linalg.inv(array)
print("Déterminant = ",determinant)
print('Inverse : ',inverse)
```

Exercice 4 : Manipulation de données avec NumPy et Pandas
Instructions

Sélection conditionnelle des données.
Fonctions d'agrégation telles que la somme et la moyenne.

```python
df = pd.DataFrame(np.random.randint(0,50,size=(10, 4)), columns=list('ABCD'))
print(df)
result_sum_pd = df[(df['A']>10) & (df['B']<20)].sum()
print(result_sum_pd)
result_mean_pd = df[(df['A']>10) & (df['B']<20)].mean()
print(result_mean_pd)
```

Exercice 6 : Tests d'hypothèses de base
Instructions
Créez un jeu de données d'exemple pour tester l'efficacité d'un nouveau programme de formation sur la productivité des employés
Étant donné un ensemble de données, formulez une hypothèse simple et testez-la à l'aide des fonctions statistiques de base de NumPy.

```python
# Score de productivité des employés avant le programme d'entrainement
productivity_before = np.random.normal(loc=50, scale=10, size=30)
print(productivity_before)

# Score de productivité des mêmes employés après le programme d'entrainement
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)
print(productivity_after)
```

```python
# T-test de l'hypothèse
stat, p = stats.ttest_ind(productivity_before, productivity_after)
alpha = 0.05
if p < alpha:
    print("L'hypothèse est rejetée. La formation n'a aucun effet sur le rendement des employés.")
else:
    print("L'hypothèse est acceptée. La formation a augmenté le rendement des employés.")
print(stat, p)
```
