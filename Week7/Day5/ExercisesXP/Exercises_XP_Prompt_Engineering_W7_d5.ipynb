## Exercise 2: Multi-Part Prompt for Quiz Generation
**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

### Tasks
- Ask for: 2-bullet summary + 3 MCQs (1 correct, 2 distractors).
- Require friendly tone, simple vocab (11?13 y/o), and Google Slides-ready output.
- Explain 3 ways your prompt improves on: `Make a quiz for kids about this article.`

```python
# TODO: write the full multi-part prompt
quiz_prompt = """
Tu es un assistant pédagogique spécialisé en sciences naturelles.

Objectif : Créer du contenu pour une présentation Google Slides sur les éruptions volcaniques destinée à des élèves de 12-13 ans (niveau 5ème).

Instructions :
1. Rédige un résumé en 2 puces (bullets points) expliquant la différence entre un volcan effusif et explosif. Utilise un vocabulaire simple.
2. Génère 3 questions à choix multiples (QCM). Pour chaque question, fournis 1 bonne réponse et 2 distracteurs plausibles.

Contraintes :
- Ton : Amical et encourageant.
- Format : Présentation claire, prête à être copiée sur des diapositives.
- Langue : Français.
"""
display(quiz_prompt)
```

```python
# TODO: explain three improvements over the vague prompt
improvements = [
    'Ciblage de l\'audience : Le prompt spécifie l\'âge (12-13 ans), ce qui ajuste automatiquement le niveau de complexité du vocabulaire.',
    'Structure de sortie forcée : En demandant explicitement des distracteurs et un résumé en puces, on évite les réponses trop génériques ou désorganisées.',
    'Contexte d\'utilisation : Mentionner \'Google Slides\' incite le modèle à produire un texte concis et visuellement structuré plutôt qu\'un long paragraphe.'
]
display(improvements)
```

## Exercise 6: Hallucination Spotting and Mitigation

### Tasks
1) Prompt the model to avoid hallucinations using only provided text.
2) Rewrite with verification language.
3) List 2 mitigation strategies.
4) Name 2 domains where hallucinations are risky and why.

```python
# TODO: baseline anti-hallucination prompt
no_hallucination_prompt = """
Résume l'article ci-dessous sur le climat.

Contrainte de vérité : Utilise UNIQUEMENT les informations présentes dans le texte fourni. Si une information n'est pas mentionnée, ne l'invente pas. Si tu as un doute, écris 'Information non disponible'.
"""
display(no_hallucination_prompt)
```

```python
# TODO: verification-focused rewrite
verification_prompt = """
1. Résume les points clés de l'article sur la biologie marine.
2. Pour chaque point cité, extrais la phrase exacte du texte source qui soutient cette affirmation.
3. Vérifie s'il existe des contradictions internes avant de répondre.
"""
display(verification_prompt)
```

```python
# TODO: list mitigation strategies and risky domains
mitigation_strategies = [
    'Utiliser le Few-Shot Prompting avec des exemples de réponses correctes et incorrectes.',
    'Demander au modèle de citer ses sources directement dans le texte fourni (Citation-based extraction).'
]
risky_domains = {
    'domain_1': {'name': 'Médical', 'why': 'Une recommandation erronée sur un traitement peut mettre des vies en danger.'},
    'domain_2': {'name': 'Juridique', 'why': 'L\'invention de lois ou de précédents peut mener à des sanctions légales graves.'},
}
display(mitigation_strategies, risky_domains)
```
