## Exercise 1: Open Source Levels Reflection
**Use:** Fully Open, Weights Released, Architecture Only; model components; fine-tuning/domain adaptation; healthcare compliance.

**Step-by-step**
1) Gather definitions for the three openness levels.  
2) Identify key characteristics (what is open; what you can/cannot do).  
3) Compare side by side in a table (what's open vs. impact).  
4) Write a 3?5 sentence comparative paragraph.  
5) Answer the healthcare prompt (1?2 sentences) about retraining on clinical data.

**Deliverables:** paragraph + healthcare answer.

```python
open_source_levels = {
    "Fully Open": {
        "definition": "Models where the training data, code, and weights are all publicly released under permissive licenses.",
        "what_is_open": ["Training data", "Weights", "Pre-training code"],
        "what_you_can_do": ["Reproduce the model", "Audit data biases", "Commercial use"],
        "what_you_cannot_do": ["Claim exclusive ownership", "Violate license restrictions"],
    },
    "Weights Released": {
        "definition": "Models where pre-trained weights are shared, but the full training data or code remains proprietary.",
        "what_is_open": ["Model weights", "Inference code"],
        "what_you_can_do": ["Fine-tune on private data", "Self-host", "Inference"],
        "what_you_cannot_do": ["Know the exact training data", "Retrain from scratch"],
    },
    "Architecture Only": {
        "definition": "Only the model's structural design (paper/code) is shared, without any learned weights.",
        "what_is_open": ["Network architecture", "Mathematical framework"],
        "what_you_can_do": ["Implement the logic", "Train your own version"],
        "what_you_cannot_do": ["Use it immediately", "Access pre-trained knowledge"],
    },
}
display(open_source_levels)
```

```python
comparison_table = """| Openness level | What's open? | Impact on retraining/modifying |
| --- | --- | --- |
| Fully Open | Data, Code, Weights | Maximum transparency; allows full replication and deep audits. |
| Weights Released | Weights & Inference Code | Fast deployment; allows fine-tuning but limits data-level transparency. |
| Architecture Only | Model structure | High barrier; requires massive compute to train from scratch. |"""
print(comparison_table)
```

```python
comparative_paragraph = """Fully open models provide the highest level of trust and flexibility by exposing the entire pipeline, including training data. Weights-released models, like Llama or Mistral, strike a balance by allowing local hosting and fine-tuning without requiring massive compute for initial training. Architecture-only releases are common in academia but offer the least immediate utility for developers. Choosing between them depends on whether the goal is absolute transparency, rapid deployment, or theoretical research."""
print(comparative_paragraph)
```

```python
healthcare_prompt_answer = """To retrain a model on clinical data while ensuring HIPAA/GDPR compliance, a 'Weights Released' or 'Fully Open' model is best. This allows for local deployment on secure servers, ensuring that sensitive patient data never leaves the controlled environment during fine-tuning."""
print(healthcare_prompt_answer)
```
