# RAPPORT DE MÉTA-ANALYSE

**Thème choisi :**  
Les techniques d’amélioration des performances des Large Language Models (LLM) : alignement, efficacité et raisonnement.

## INTRODUCTION

Depuis l’apparition des grands modèles de langages (LLM), ceux-ci ont profondément transformé le domaine du traitement automatique du langage naturel. Grâce à leur capacité à comprendre et générer du texte, ils sont aujourd’hui utilisés dans de nombreux domaines tels que l’assistance virtuelle, la programmation, l’éducation ou la recherche scientifique.

Toutefois, plusieurs défis subsistent : améliorer leurs performances, réduire leurs coûts de calcul, mieux les aligner avec les préférences humaines et renforcer leur capacité de raisonnement.

Cette méta-analyse a pour objectif d’étudier quatre travaux récents qui proposent différentes solutions à ces défis. Les articles analysés abordent respectivement :
1. L’optimisation du fine-tuning (**QLoRA**).
2. L’alignement des modèles (**Direct Preference Optimization**).
3. L’amélioration du raisonnement (**Self-Refine**).
4. La conception d’un modèle de nouvelle génération (**Llama 3**).

Les publications proviennent principalement de arXiv et des équipes de Meta.

## A. Résumés des documents

### Article 1 : [QLora : Efficient Finetuning of Quantiqed LLMs](https://arxiv.org/abs/2305.14314)


*   **Citation :** Dettmers T., et al. (2023). QLoRA: Efficient Finetuning of Quantized LLMs. arXiv.
*   **Problème :** Le fine-tuning complet d’un LLM nécessite énormément de mémoire GPU.
*   **Solution :** Les auteurs proposent QLoRA. Le modèle est quantifié en 4 bits tandis que seules les matrices LoRA sont entraînées.
*   **Jeu de données :** Alpaca, OASST1, Vicuna benchmark.
*   **Architecture :** LLaMA, LoRA, Quantification NF4.
*   **Métriques :** Vicuna Evaluation, MMLU, mémoire GPU.

**Résultat :** Il est possible de fine-tuner un modèle de 65 milliards de paramètres sur une seule carte graphique de 48 Go tout en conservant des performances proches du fine-tuning classique.

### Article 2 : Direct Preference Optimization : Your Language Model is Secretly a Reward Model

*   **Citation :** Rafailov, R., Sharma, A., Mitchell, E., et al. (2023). Direct Preference Optimization. arXiv.
*   **Problème :** L'alignement des LLM via RLHF est complexe, instable et nécessite l'entraînement d'un modèle de récompense séparé.
*   **Solution :** DPO (Direct Preference Optimization) permet d'optimiser directement le modèle de langage sur les préférences humaines sans passer par un modèle de récompense ou l'apprentissage par renforcement complexe.
*   **Impact :** Simplification considérable du pipeline d'alignement tout en obtenant des performances égales ou supérieures au RLHF.

### Article 3 : Self-Refine: Iterative Refinement with Self-Feedback

*   **Citation :** Madaan, A., et al. (2023). Self-Refine: Iterative Refinement with Self-Feedback. arXiv.
*   **Problème :** Les LLM génèrent souvent des réponses imparfaites, contenant des erreurs logiques ou de style, lors de la première tentative.
*   **Solution :** Une approche itérative où le même modèle génère une sortie, fournit un feedback sur sa propre réponse, puis l'affine en fonction de ce feedback.
*   **Caractéristique :** Ne nécessite aucun entraînement supplémentaire ni données d'alignement humain.
*   **Impact :** Amélioration significative sur des tâches de raisonnement, de génération de code et de dialogue.

### Article 4 : Llama 3: Open Foundation and Fine-Tuned Chat Models

*   **Citation :** Meta AI (2024). Introducing Llama 3.
*   **Problème :** Le besoin de modèles open-source performants capables de rivaliser avec les modèles propriétaires fermés.
*   **Solution :** Un modèle entraîné sur un corpus massif de 15 trillions de tokens avec une architecture transformeur optimisée.
*   **Architecture :** Tokenizer plus efficace (128k vocabulaire), Grouped Query Attention (GQA) pour toutes les tailles.
*   **Impact :** Établit un nouvel état de l'art pour les modèles ouverts en termes de raisonnement, de connaissances et de suivi d'instructions.

## C. Perspectives

Plusieurs tendances apparaissent clairement :

1.  **Réduction des coûts :** Les chercheurs cherchent désormais à réduire le coût d'entraînement. Les techniques de quantification et de fine-tuning léger deviennent la norme.
2.  **Alignement :** Les méthodes comme DPO remplacent progressivement des approches plus complexes comme RLHF, simplifiant l'entraînement.
3.  **Raisonnement :** L'amélioration ne repose plus uniquement sur la taille des modèles. Des approches comme *Self-Refine* montrent qu'un modèle peut s'améliorer via un processus itératif.
4.  **Modèles Ouverts :** Développement de modèles performants et accessibles comme Llama 3.

## Conclusion

Cette méta-analyse montre que les recherches s'orientent vers l'efficacité, la réduction des coûts, l'alignement utilisateur et le raisonnement avancé. Les progrès reposent désormais sur des innovations de stratégies d'optimisation plutôt que sur la simple augmentation du nombre de paramètres.
